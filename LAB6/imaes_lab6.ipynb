# **Experiment 6: Fuzzy Logic and Constraint Satisfaction**

```python
import numpy as np
import matplotlib.pyplot as plt

# Triangular membership function
def triangular(x, a, b, c):
    if x < a or x > c:
        return 0.0
    if x == b:
        return 1.0
    if x < b:
        return (x - a) / (b - a) if b != a else 1.0
    return (c - x) / (c - b) if c != b else 1.0

def left_shoulder(x, a, b):
    if x <= a:
        return 1.0
    if x >= b:
        return 0.0
    return (b - x) / (b - a)

def right_shoulder(x, a, b):
    if x <= a:
        return 0.0
    if x >= b:
        return 1.0
    return (x - a) / (b - a)

# Membership functions for attendance and marks
def attendance_membership(value):
    return {
        "low": left_shoulder(value, 30, 60),
        "medium": triangular(value, 30, 60, 90),
        "high": right_shoulder(value, 60, 90)
    }

def marks_membership(value):
    return {
        "low": left_shoulder(value, 30, 50),
        "medium": triangular(value, 30, 50, 80),
        "high": right_shoulder(value, 50, 80)
    }

# Output membership functions
def performance_membership(value):
    return {
        "poor": left_shoulder(value, 25, 50),
        "average": triangular(value, 25, 50, 75),
        "good": right_shoulder(value, 50, 75)
    }

# Fuzzy inference using simple rule aggregation and weighted average
def predict_performance(attendance, marks):
    A = attendance_membership(attendance)
    M = marks_membership(marks)

    rules = [
        (min(A["low"], 1.0), "poor"),
        (min(M["low"], 1.0), "poor"),
        (min(A["medium"], M["medium"]), "average"),
        (min(A["high"], M["high"]), "good"),
        (min(A["high"], M["medium"]), "average"),
        (min(A["medium"], M["high"]), "good"),
        (min(A["low"], M["high"]), "average")
    ]

    # Representative output values
    output_values = {"poor": 25, "average": 50, "good": 85}

    weighted_sum = sum(
        strength * output_values[label]
        for strength, label in rules
    )
    total_strength = sum(strength for strength, _ in rules)

    if total_strength == 0:
        return 50.0

    return weighted_sum / total_strength

# Test different inputs
test_cases = [
    (20, 25),
    (50, 55),
    (90, 90),
    (40, 85),
    (80, 60)
]

for attendance, marks in test_cases:
    performance = predict_performance(attendance, marks)
    print(
        f"Attendance: {attendance}%, Marks: {marks} "
        f"-> Performance: {performance:.2f}"
    )
```

```python
# Variables: subjects
subjects = ["AI", "DBMS", "Mathematics"]

# Domains: possible time slots
domains = {
    "AI": ["Mon 9 AM", "Mon 10 AM", "Tue 9 AM"],
    "DBMS": ["Mon 9 AM", "Mon 10 AM", "Tue 9 AM"],
    "Mathematics": ["Mon 9 AM", "Mon 10 AM", "Tue 9 AM"]
}

timetable = {}

def get_day(slot):
    return slot.split()[0]

def is_valid(subject, slot):
    for assigned_subject, assigned_slot in timetable.items():

        # Constraint 1: no two subjects in the same slot
        if assigned_slot == slot:
            return False

        # Constraint 2: AI and DBMS must be on different days
        if {subject, assigned_subject} == {"AI", "DBMS"}:
            if get_day(slot) == get_day(assigned_slot):
                return False

    return True

def backtrack(index=0):
    if index == len(subjects):
        return True

    subject = subjects[index]

    for slot in domains[subject]:
        if is_valid(subject, slot):
            timetable[subject] = slot

            if backtrack(index + 1):
                return True

            # Backtrack if this assignment leads to no solution
            del timetable[subject]

    return False

if backtrack():
    print("Valid Timetable:")
    for subject, slot in timetable.items():
        print(f"{subject}: {slot}")
else:
    print("No valid timetable found.")
```
